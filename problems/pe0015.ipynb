<p>How many such routes are there through a $20 \times 20$ grid?</p>



```python
#this is really just pascal's triangle where you only look at the middle end point
from math import factorial

#for a 20x20 grid, the number of routes is given by the central binomial coefficient
print(factorial(40) // (factorial(20) * factorial(20)))
```
