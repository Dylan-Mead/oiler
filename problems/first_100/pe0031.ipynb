<p>In the United Kingdom the currency is made up of pound (£) and pence (p). There are eight coins in general circulation:</p>
<blockquote>1p, 2p, 5p, 10p, 20p, 50p, £1 (100p), and £2 (200p).</blockquote>
<p>It is possible to make £2 in the following way:</p>
<blockquote>1×£1 + 1×50p + 2×20p + 1×5p + 1×2p + 3×1p</blockquote>
<p>How many different ways can £2 be made using any number of coins?</p>



```python
coins = [1, 2, 5, 10, 20, 50, 100, 200]

principle = 200
```

```python
#This method works by using dynamic programming to count the number of ways to make change for a given amount with a set of coins.
#The function uses a bottom-up approach to build an array where each index represents the number of ways to make change for that amount.
#This approach ensures that each combination is counted only once, avoiding double counting.
#https://www.geeksforgeeks.org/dsa/coin-change-dp-7/
def count_ways_to_make_change(principle, coins):
    ways = [0] * (principle + 1)
    ways[0] = 1
    for coin in coins:
        for i in range(coin, principle + 1):
            ways[i] += ways[i - coin]
    return ways[principle]
```

```python
for coin in coins:
    print('Ammount:',coin, ', Ways:', count_ways_to_make_change(coin, coins))
```
