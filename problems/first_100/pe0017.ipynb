If the numbers $1$ to $5$ are written out in words: one, two, three, four, five, then there are $3 + 3 + 5 + 4 + 4 = 19$ letters used in total.
If all the numbers from $1$ to $1000$ (one thousand) inclusive were written out in words, how many letters would be used? 
NOTE: Do not count spaces or hyphens. For example, $342$ (three hundred and forty-two) contains $23$ letters and $115$ (one hundred and fifteen) contains $20$ letters. The use of "and" when writing out numbers is in compliance with British usage.


```python
#This is pretty annoying. First write a dictionary mapping numbers to their word representations
num_to_words = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five",
    6: "six",
    7: "seven",
    8: "eight",
    9: "nine",
    10: "ten",
    11: "eleven",
    12: "twelve",
    13: "thirteen",
    14: "fourteen",
    15: "fifteen",
    16: "sixteen",
    17: "seventeen",
    18: "eighteen",
    19: "nineteen",
    20: "twenty",
    30: "thirty",
    40: "forty",
    50: "fifty",
    60: "sixty",
    70: "seventy",
    80: "eighty",
    90: "ninety",
    100: "one hundred",
    200: "two hundred",
    300: "three hundred",
    400: "four hundred",
    500: "five hundred",
    600: "six hundred",
    700: "seven hundred",
    800: "eight hundred",
    900: "nine hundred",
    1000: "one thousand"
} #thank you copilot

#then a dictionary mapping numbers to their char count
num_to_char_count = {num: len(word.replace(" ", "")) for num, word in num_to_words.items()}

```

```python
def number_to_char_count(n):
    string_number = str(n)
    if n in num_to_char_count:
        return num_to_char_count[n]
    else:
        if n < 100:
            tens, ones = divmod(n, 10)
            return num_to_char_count[tens * 10] + num_to_char_count[ones]
        elif n < 1000:
            hundreds, remainder = divmod(n, 100)
            if remainder == 0:
                return num_to_char_count[hundreds * 100]
            else:
                return num_to_char_count[hundreds * 100] + 3 + number_to_char_count(remainder)  # 3 for "and"
        else:
            raise ValueError("Number out of range")
    
```

```python
total = sum(number_to_char_count(n) for n in range(1, 1001))
total
```
