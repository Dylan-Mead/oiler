The Fibonacci sequence is defined by the recurrence relation:
$F_n = F_{n - 1} + F_{n - 2}$, where $F_1 = 1$ and $F_2 = 1$.
Hence the first $12$ terms will be:
$$\begin{align}
F_1 &amp;= 1\\
F_2 &amp;= 1\\
F_3 &amp;= 2\\
F_4 &amp;= 3\\
F_5 &amp;= 5\\
F_6 &amp;= 8\\
F_7 &amp;= 13\\
F_8 &amp;= 21\\
F_9 &amp;= 34\\
F_{10} &amp;= 55\\
F_{11} &amp;= 89\\
F_{12} &amp;= 144
\end{align}$$
The $12^{th}$ term, $F_{12}$, is the first term to contain three digits.
What is the index of the first term in the Fibonacci sequence to contain $1000$ digits?



```python
#We could get to 1000 digits pretty quickly but it'll be more fun to use a binary search algorithm to find the first Fibonacci number with 1000 digits.
import numpy as np
#create a fibonacci sequence generator using matrix exponentiation

#We will store the even powers of the Fibonacci matrix to use them in a binary search algorithm later.
#Going to create some veryt large ints so we need the numpy object type
fib_generator = np.array([[0, 1],
                          [1, 1]], dtype=object)

even_powers_of_fibonaci = [fib_generator]
power = 1
fib_power = fib_generator

while len(str(fib_power[0, 1])) < 1000:
    #print(fib_power)
    fib_power = fib_power @ fib_power
    even_powers_of_fibonaci.append(fib_power)

```

```python
#The last one in the list of even powers of Fibonacci numbers is too big so we start with the second to last one
index_power = len(even_powers_of_fibonaci) - 2#because of zero indexing
index_powers = [index_power]
start_fib = even_powers_of_fibonaci[index_power]
while index_power > 0:
    index_power -= 1
    if len(str((start_fib @ even_powers_of_fibonaci[index_power])[0, 1])) < 1000:
        start_fib = start_fib @ even_powers_of_fibonaci[index_power]
        index_powers.append(index_power)
        print(index_powers)
```

```python
#this little loop calculates the index of the last Fibonacci number with 999 digits
#basically what we have is a binary number that we are converting to base 10
index = 0
for power in index_powers:
    index += 2**power

print(index + 1) #we calculated the index of the last term with 999 digits so we have to add one
print(len(str(start_fib[0, 1])))
print(len(str(start_fib[1, 1])))
```
