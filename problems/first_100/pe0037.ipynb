The number $3797$ has an interesting property. Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797$, $797$, $97$, and $7$. Similarly we can work from right to left: $3797$, $379$, $37$, and $3$.
Find the sum of the only eleven primes that are both truncatable from left to right and right to left.
NOTE: $2$, $3$, $5$, and $7$ are not considered to be truncatable primes.



```python
#first collect our primes below 1,000,000
with open("../../src/oiler/primes.txt", "r") as f:
    primes = [int(line.strip()) for line in f if int(line.strip()) < 1000000 ]#[4:]
```

```python
def is_truncatable_prime(n, primes):

    str_n = str(n)
    if (set(str_n) & set('0468')):
        return False
    for i in range(1, len(str_n)):
        if int(str_n[i:]) not in primes or int(str_n[:i]) not in primes:
            return False
    return True

truncatable_primes = [p for p in primes[4:] if is_truncatable_prime(p, primes)]
print(truncatable_primes)
print(len(truncatable_primes))
print(sum(truncatable_primes))
```
