$145$ is a curious number, as $1! + 4! + 5! = 1 + 24 + 120 = 145$.
Find the sum of all numbers which are equal to the sum of the factorial of their digits.
Note: As $1! = 1$ and $2! = 2$ are not sums they are not included.


```python
from math import factorial

curious_numbers = []
for i in range(10, 150):
    digits = [int(d) for d in str(i)]
    digit_sum = 0
    for digit in digits:
        digit_sum += factorial(digit)
    if digit_sum == i:
        curious_numbers.append(i)

curious_numbers
```

```python
def find_curious_numbers(curious_numbers):
    from math import factorial
    start = max(curious_numbers, default=10) + 1
    for i in range(start, 3628800):
        digits = [int(d) for d in str(i)]
        digit_sum = 0
        for digit in digits:
            digit_sum += factorial(digit)
        if digit_sum == i:
            curious_numbers.append(i)
    return curious_numbers
```

```python
start_length = len(curious_numbers)
end_length = len(curious_numbers) + 1
while start_length < end_length:
    curious_numbers = find_curious_numbers(curious_numbers)
    start_length = end_length
    end_length = len(curious_numbers)

curious_numbers
```

```python
sum(curious_numbers)
```
