The following iterative sequence is defined for the set of positive integers:

$n \to n/2$ ($n$ is even)

$n \to 3n + 1$ ($n$ is odd)

Using the rule above and starting with $13$, we generate the following sequence:
$$13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1.$$
It can be seen that this sequence (starting at $13$ and finishing at $1$) contains $10$ terms. Although it has not been proved yet (Collatz Problem), it is thought that all starting numbers finish at $1$.
Which starting number, under one million, produces the longest chain?
Once the chain starts the terms are allowed to go above one million.



```python
def collatz_sequence(n):
    sequence = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    return sequence
```

```python
#We can start with a list of every number from one to 999,999 but remove numbers as they come up in other sequences to avoid redundant calculations.
#for example: 1 -> 4 -> 2 -> 1 means that the sequence length for 4 and 2 is shorter than the sequence length for 1 and need not be checked separately.

top_number = 1
top_length = 1
top_sequence = [1]
```

```python
collatz_candidates = set(range(1, 1000000))

while len(collatz_candidates) > 0:
    i = collatz_candidates.pop()
    cs = collatz_sequence(i)
    if len(cs) > top_length:
        top_length = len(cs)
        top_number = i
        top_sequence = cs
    collatz_candidates -= set(cs)

print(top_number, top_length)
```
