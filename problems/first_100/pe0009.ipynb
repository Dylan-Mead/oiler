A Pythagorean triplet is a set of three natural numbers, $a \lt b \lt c$, for which,
$$a^2 + b^2 = c^2.$$
For example, $3^2 + 4^2 = 9 + 16 = 25 = 5^2$.
There exists exactly one Pythagorean triplet for which $a + b + c = 1000$. Find the product $abc$.



```python
#Doing some quick algebra we get: a = (500000 - 1000b) / (1000 - b)
#Adding the constraint that a and b must be integers we can guess a and check if b is an integer and less than 1000
```

```python
def derive_a(b):
    return (500000 - 1000*b) / (1000 - b)

def derive_c(a, b):
    return 1000 - a - b

for b in range(1, 1000):
    a = derive_a(b)
    if a.is_integer() and a < 1000 and a > 0:
        a = int(a)
        c = derive_c(a, b)
        if c > 0:
            print(a, b, c, a*b*c)
            break
```
