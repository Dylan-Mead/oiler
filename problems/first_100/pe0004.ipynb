A palindromic number reads the same both ways. The largest palindrome made from the product of two $2$-digit numbers is $9009 = 91 \times 99$.
Find the largest palindrome made from the product of two $3$-digit numbers.



```python
#Starting with the largest product of two 3-digit numbers and working down.
#Avoiding duplicate product checks
found = False
for i in range(1998, 0, -1):
    if found:
        break
    b = (i+1) // 2
    a = i - b
    #print(a, b, i)
    product = a * b
    if str(product) == str(product)[::-1]:
        print('palindrome', a, b, product)
        found = True
        break
    while b < 999:
        b += 1
        a -= 1
        #print(a, b, i)
        product = a * b
        if str(product) == str(product)[::-1]:
            print('palindrome', a, b, product)
            found = True
            break
```
