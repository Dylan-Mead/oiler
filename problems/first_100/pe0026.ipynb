A unit fraction contains $1$ in the numerator. The decimal representation of the unit fractions with denominators $2$ to $10$ are given:
$$\begin{align}
1/2 &amp;= 0.5\\
1/3 &amp;=0.(3)\\
1/4 &amp;=0.25\\
1/5 &amp;= 0.2\\
1/6 &amp;= 0.1(6)\\
1/7 &amp;= 0.(142857)\\
1/8 &amp;= 0.125\\
1/9 &amp;= 0.(1)\\
1/10 &amp;= 0.1
\end{align}$$
Where $0.1(6)$ means $0.166666\cdots$, and has a $1$-digit recurring cycle. It can be seen that $1/7$ has a $6$-digit recurring cycle.
Find the value of $d \lt 1000$ for which $1/d$ contains the longest recurring cycle in its decimal fraction part.



```python
def get_repeating_cycle_length(n: int) -> int:

    #first remove factors of 2 and five because they do not contribute to the repeating cycle
    while n % 2 == 0:
        n //= 2
    while n % 5 == 0:
        n //= 5
    if n == 1:
        return 0
    
    #This part calculates the length of the repeating cycle for the remaining fraction 1/n
    remainder = 1
    length = 0
    while True:
        remainder = (remainder * 10) % n
        length += 1
        if remainder == 1:
            return length
```

```python
#We can reduce the amount of numbers to check by taking advantage of the fact that the max length of a repeating cycle for 1/n is less than n - 1.
#So what we'll do is only check primes less than 1000 and stop when we get to one with cycle length n - 1.

with open("../../src/oiler/primes.txt") as f:
    primes = [int(line.strip()) for line in f if int(line.strip()) < 1000][::-1]

longest_cycle = 0
number_with_longest_cycle = 0

for n in primes:
    cycle_length = get_repeating_cycle_length(n)
    if cycle_length > longest_cycle:
        longest_cycle = cycle_length
        number_with_longest_cycle = n
        print(f"The number under 1000 with the longest repeating cycle is {number_with_longest_cycle} with a cycle length of {longest_cycle}")
    if cycle_length == n - 1:
        break
```
