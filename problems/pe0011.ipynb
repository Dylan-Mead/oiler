```python
sa = [[int(y) for y in x.split(' ')] for x in s.splitlines()]
sa
```

```python
for row in sa:
    print(row)
```

```python
max_product = 0

#we'll check horizontal products first, then, vertical, then diagonals

#horizontal checks
for row in sa:
    for i in range(len(row) - 3):
        product = row[i] * row[i+1] * row[i+2] * row[i+3]
        if product > max_product:
            max_product = product

print(max_product)
```

```python
#vertical checks
for col in range(len(sa[0])):
    for row in range(len(sa) - 3):
        product = sa[row][col] * sa[row+1][col] * sa[row+2][col] * sa[row+3][col]
        if product > max_product:
            max_product = product

print(max_product)
```

```python
#diagonal right to left checks
for row in range(len(sa) - 3):
    for col in range(3, len(sa[0])):
        product = sa[row][col] * sa[row+1][col-1] * sa[row+2][col-2] * sa[row+3][col-3]
        if product > max_product:
            max_product = product

print(max_product)
```

```python
#diagonal left to right checks
for row in range(len(sa) - 3):
    for col in range(len(sa[0]) - 3):
        product = sa[row][col] * sa[row+1][col+1] * sa[row+2][col+2] * sa[row+3][col+3]
        if product > max_product:
            max_product = product

print(max_product)
```
